```python
import replicate
from IPython.display import Image, display
from replicate import Client
```

```python
MODEL_NAME = "black-forest-labs/flux-1.1-pro"
prompt =  "Dark beutiful Jungle in Canada where a happy girl with 90s style is standing beside a bear and hold a paper which contains: FLUX 1 . 1 Pro Abtin Zandi"
aspect_ratio = "1:1"
format = "webp"
```

```python
from google.colab import userdata

REPLICATE_API_TOKEN = userdata.get('REPLICATE_API_TOKEN')

# Ensure the token was retrieved successfully
if not REPLICATE_API_TOKEN:
    raise ValueError("REPLICATE_API_TOKEN not found in Colab secrets. Please add it.")
else:
    print("Replicate API token retrieved successfully from Colab secrets.")
```

```python
client = Client(api_token=REPLICATE_API_TOKEN)

output = client.run(
    MODEL_NAME,
    input={
        "prompt": prompt,
        "aspect_ratio": aspect_ratio,
        "output_format": format,
        "output_quality": 80,
        "safety_tolerance": 2,
        "prompt_upsampling": True

    }
)
```

```python
print("Final Result of flux-kontext-pro image generation process\n",output)
```

```python
with open("my-image.png", "wb") as file:
    file.write(output.read())

display(Image(url=output.url))
```
