##**Google - Progan/128**


Image generator based on tensorflow reimplementation of Progressive GANs.

Maps from a 512-dimensional latent space to images. During training, the latent space vectors were sampled from a normal distribution.

```python
import tensorflow as tf
import tensorflow_hub as hub
import numpy as np
import PIL.Image
```

```python
face_generator = hub.load("https://kaggle.com/models/google/progan-128/tensorflow1/progan-128/1")
```

```python
latent_vector = np.random.normal(size=[1,512]).astype(np.float32)

image = face_generator.signatures['default'](tf.constant(latent_vector))['default']
```

```python
PIL.Image.fromarray(np.uint8(image.numpy()[0]*255))
```
