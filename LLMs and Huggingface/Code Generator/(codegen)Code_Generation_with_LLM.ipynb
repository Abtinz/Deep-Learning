##**Salesforce - CodeGen**

CodeGen is a family of autoregressive language models for program synthesis from the paper: A Conversational Paradigm for Program Synthesis by Erik Nijkamp, Bo Pang, Hiroaki Hayashi, Lifu Tu, Huan Wang, Yingbo Zhou, Silvio Savarese, Caiming Xiong. The models are originally released under 3 pre-training data variants (NL, Multi, Mono) and 4 model size variants (350M, 2B, 6B, 16B).

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
```

##Load the model, tockenizer, and pipeline from hugginface

```python
tokenizer = AutoTokenizer.from_pretrained("Salesforce/codegen-2B-multi")
model = AutoModelForCausalLM.from_pretrained("Salesforce/codegen-2B-multi")
```

```python
code_generator = pipeline("text-generation", model="Salesforce/codegen-2B-multi")
```

```python
def generate_code(prompt):
    """
    Generate code based on the provided prompt using the code_generator pipeline.

    Args:
        prompt (str): The input prompt for generating code.

    Returns:
        str: The generated code.
    """

    if prompt is None :
      prompt = input("Enter a coding task (e.g., 'Write a Python function to ping google.com'):")

    print(f"*************\nusers prompt:\n{prompt}")
    response = code_generator(prompt,
                              max_length=512, # Set the maximum length of the generated code.
                              truncation=True, # Truncate the input if it's longer than the maximum length.
                              num_return_sequences=1, # Generate only one sequence of code.
                              do_sample=True, # Use sampling for code generation to introduce randomness.
                              temperature=0.7, # Control the randomness of the output. Higher values mean more randomness.
                              top_k=50, # Consider only the top_k most likely tokens at each step.
                              top_p=0.95) # Consider tokens with a cumulative probability of top_p.
    print("\n*************Generated Code:\n")

    # Extract the generated text from the response.
    print(response[0]['generated_text'])
```

```python
generate_code(
    prompt = """
      Write a Python code to create a request using requests library to google.com
    """
)
```
